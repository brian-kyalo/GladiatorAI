```python
import pandas as pd


# ============================================================
# GLADIATOR AI — SINGLE FIGHT PIPELINE VALIDATION
# ============================================================

# 1. Create the Dataset Builder
builder = DatasetBuilder(df)

# 2. Select ONE historical fight
row_index = 100
fight = df.iloc[row_index]

fight_date = pd.to_datetime(fight["date"])

print("=" * 70)
print("GLADIATOR AI — SINGLE FIGHT VALIDATION")
print("=" * 70)

print("\nSOURCE FIGHT")
print("-" * 70)

print(f"Row index : {row_index}")
print(f"Date      : {fight_date.date()}")
print(f"Red       : {fight['R_fighter']}")
print(f"Blue      : {fight['B_fighter']}")
print(f"Winner    : {fight['Winner']}")
print(f"Finish    : {fight['finish']}")


# ============================================================
# 3. Build ONE matchup
# ============================================================

matchup, skipped = builder.build_fight(
    row_index=row_index,
    fight=fight,
)


# ============================================================
# 4. Validate skipped status
# ============================================================

print("\nBUILD STATUS")
print("-" * 70)

print("Skipped:", skipped)

if skipped is not None:

    print("\n❌ FIGHT WAS SKIPPED")
    print("Reason:", skipped.reason)

else:

    print("✅ Fight successfully converted into a matchup")


# ============================================================
# 5. Stop immediately if no matchup was created
# ============================================================

if matchup is None:

    raise RuntimeError(
        "No MatchupProfile was created. "
        "The fight was skipped."
    )


# ============================================================
# 6. Print MatchupProfile
# ============================================================

print("\nMATCHUP PROFILE")
print("-" * 70)

print(matchup)


# ============================================================
# 7. Convert to dictionary
# ============================================================

matchup_dict = builder.matchup_engine.to_dict(
    matchup
)

print("\nMATCHUP DICTIONARY")
print("-" * 70)

print(matchup_dict)


# ============================================================
# 8. Convert to one-row DataFrame
# ============================================================

matchup_df = builder.matchup_engine.to_dataframe(
    matchup
)

print("\nMATCHUP DATAFRAME")
print("-" * 70)

print(matchup_df)

print("\nDataFrame shape:", matchup_df.shape)


# ============================================================
# 9. Validate the expected one-row structure
# ============================================================

assert len(matchup_df) == 1, (
    "Expected exactly one ML row."
)

assert matchup.red_fighter == fight["R_fighter"], (
    "Red fighter mismatch."
)

assert matchup.blue_fighter == fight["B_fighter"], (
    "Blue fighter mismatch."
)

assert matchup.winner == fight["Winner"], (
    "Winner mismatch."
)

assert matchup.snapshot_date == str(
    fight_date.date()
), (
    "Snapshot date mismatch."
)


# ============================================================
# 10. LEAKAGE TEST
# ============================================================

print("\nLEAKAGE CHECK")
print("-" * 70)

red_history = builder.history_engine.get_history(
    fighter=fight["R_fighter"],
    before_date=fight_date,
)

blue_history = builder.history_engine.get_history(
    fighter=fight["B_fighter"],
    before_date=fight_date,
)


# Red history must contain only earlier fights
if not red_history.empty:

    red_latest_date = pd.to_datetime(
        red_history["date"].iloc[0]
    )

    print(
        "Red latest historical fight:",
        red_latest_date.date()
    )

    assert red_latest_date < fight_date, (
        "LEAKAGE DETECTED: "
        "Red history contains the current fight."
    )

else:

    print("Red has no historical fights before this matchup.")


# Blue history must contain only earlier fights
if not blue_history.empty:

    blue_latest_date = pd.to_datetime(
        blue_history["date"].iloc[0]
    )

    print(
        "Blue latest historical fight:",
        blue_latest_date.date()
    )

    assert blue_latest_date < fight_date, (
        "LEAKAGE DETECTED: "
        "Blue history contains the current fight."
    )

else:

    print("Blue has no historical fights before this matchup.")


# ============================================================
# 11. Feature sanity checks
# ============================================================

print("\nFEATURE CHECK")
print("-" * 70)

feature_columns = [
    "age_diff",
    "height_diff",
    "reach_diff",
    "experience_diff",
    "wins_diff",
    "losses_diff",
    "win_rate_diff",
    "win_streak_diff",
    "lose_streak_diff",
    "momentum_diff",
    "ko_rate_diff",
    "submission_rate_diff",
    "decision_rate_diff",
    "fight_time_diff",
    "sig_strike_diff",
    "takedown_diff",
    "submission_attempt_diff",
]

for column in feature_columns:

    value = matchup_df.iloc[0][column]

    print(f"{column:28} {value}")

    assert pd.notna(value), (
        f"Feature '{column}' is missing."
    )


# ============================================================
# 12. FINAL RESULT
# ============================================================

print("\n" + "=" * 70)
print("✅ SINGLE-FIGHT PIPELINE VALIDATION PASSED")
print("=" * 70)

print(
    "\nOne historical UFC fight successfully became "
    "one leakage-checked ML training row."
)

print(
    "\nReady for full DatasetBuilder.build()."
)
```

```python
import pandas as pd

from feature_engine.history import HistoryEngine
from feature_engine.calculators.physical import PhysicalEngine
from feature_engine.models import PhysicalStats


# ============================================================
# GLADIATOR AI — PHYSICAL ENGINE VALIDATION
# ============================================================

history_engine = HistoryEngine(df)

# ------------------------------------------------------------
# Find a real fighter from the dataset
# ------------------------------------------------------------

red_fighters = df["R_fighter"].dropna().astype(str)
blue_fighters = df["B_fighter"].dropna().astype(str)

all_fighters = pd.concat(
    [red_fighters, blue_fighters],
    ignore_index=True,
)

fighter_counts = all_fighters.value_counts()

fighter = fighter_counts[
    fighter_counts >= 2
].index[0]


# ------------------------------------------------------------
# Get this fighter's fights
# ------------------------------------------------------------

fighter_fights = df[
    (df["R_fighter"] == fighter)
    | (df["B_fighter"] == fighter)
].copy()

fighter_fights["date"] = pd.to_datetime(
    fighter_fights["date"]
)

fighter_fights = fighter_fights.sort_values(
    "date"
)


# ------------------------------------------------------------
# Choose a snapshot after their previous fight
# ------------------------------------------------------------

snapshot_date = fighter_fights["date"].iloc[-1]

history = history_engine.get_history(
    fighter=fighter,
    before_date=snapshot_date,
)


# ------------------------------------------------------------
# Build physical profile
# ------------------------------------------------------------

physical_engine = PhysicalEngine()

physical = physical_engine.build(
    history=history,
    fighter=fighter,
)


# ============================================================
# RESULTS
# ============================================================

print("=" * 70)
print("GLADIATOR AI — PHYSICAL ENGINE VALIDATION")
print("=" * 70)

print(f"\nFighter       : {fighter}")
print(f"Snapshot date : {snapshot_date.date()}")
print(f"History rows  : {len(history)}")


print("\nLatest Historical Fight")
print("-" * 70)

if not history.empty:
    print(
        history[
            [
                "date",
                "R_fighter",
                "B_fighter",
            ]
        ].iloc[0]
    )


print("\nPhysical Profile")
print("-" * 70)

print(physical)


# ============================================================
# VALIDATION
# ============================================================

print("\nValidation")
print("-" * 70)

assert not history.empty, (
    "HistoryEngine returned no historical fights."
)

assert isinstance(
    physical,
    PhysicalStats,
), (
    "PhysicalEngine did not return PhysicalStats."
)

print("✅ HistoryEngine returned real historical data.")

print("✅ PhysicalEngine returned a valid PhysicalStats object.")

print("✅ Physical validation passed.")
```

```python
import pandas as pd
import numpy as np


# ============================================================
# GLADIATOR AI — PHYSICAL DATA QUALITY AUDIT
# ============================================================

physical_columns = [
    "R_age",
    "B_age",
    "R_Height_cms",
    "B_Height_cms",
    "R_Reach_cms",
    "B_Reach_cms",
]

print("=" * 80)
print("GLADIATOR AI — RAW PHYSICAL DATA AUDIT")
print("=" * 80)


# ------------------------------------------------------------
# 1. Missing values in the RAW dataset
# ------------------------------------------------------------

print("\nRAW MISSING VALUES")
print("-" * 80)

raw_missing = df[physical_columns].isna().sum()

print(raw_missing)


# ------------------------------------------------------------
# 2. Zero values in the RAW dataset
# ------------------------------------------------------------

print("\nRAW ZERO VALUES")
print("-" * 80)

raw_zeros = (
    df[physical_columns] == 0
).sum()

print(raw_zeros)


# ------------------------------------------------------------
# 3. Extremely suspicious physical values
# ------------------------------------------------------------

print("\nSUSPICIOUS RAW VALUES")
print("-" * 80)

for column in physical_columns:

    values = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    suspicious = df[
        (values <= 0)
        |
        (values > 250)
    ]

    if not suspicious.empty:

        print(
            f"\n{column}: "
            f"{len(suspicious)} suspicious rows"
        )

        print(
            suspicious[
                [
                    "date",
                    "R_fighter",
                    "B_fighter",
                    column,
                ]
            ].head(10)
        )


# ------------------------------------------------------------
# 4. Missingness by physical category
# ------------------------------------------------------------

print("\nPHYSICAL MISSINGNESS BY CATEGORY")
print("-" * 80)

print(
    "Age missing:",
    df[
        ["R_age", "B_age"]
    ].isna().any(axis=1).sum()
)

print(
    "Height missing:",
    df[
        ["R_Height_cms", "B_Height_cms"]
    ].isna().any(axis=1).sum()
)

print(
    "Reach missing:",
    df[
        ["R_Reach_cms", "B_Reach_cms"]
    ].isna().any(axis=1).sum()
)


# ------------------------------------------------------------
# 5. Check whether zero is being used as missing
# ------------------------------------------------------------

print("\nZERO-AS-MISSING CHECK")
print("-" * 80)

for column in [
    "R_Height_cms",
    "B_Height_cms",
    "R_Reach_cms",
    "B_Reach_cms",
]:

    zero_count = (
        pd.to_numeric(
            df[column],
            errors="coerce"
        )
        .eq(0)
        .sum()
    )

    missing_count = df[column].isna().sum()

    print(
        f"{column:18}"
        f" missing={missing_count:<6}"
        f" zero={zero_count}"
    )


print("\n" + "=" * 80)
print("AUDIT COMPLETE")
print("=" * 80)
```

```python
# ============================================================
# TRACE THE SOURCE DATA FOR EACH BAD MATCHUP
# ============================================================

history_engine = HistoryEngine(df)
physical_engine = PhysicalEngine()

for _, bad_row in bad.iterrows():

    red = bad_row["red_fighter"]
    blue = bad_row["blue_fighter"]
    fight_date = pd.to_datetime(
        bad_row["snapshot_date"]
    )

    print("\n" + "=" * 80)
    print(f"{red} vs {blue} | {fight_date.date()}")
    print("=" * 80)

    # Find the actual source fight
    source = df[
        (df["R_fighter"] == red)
        & (df["B_fighter"] == blue)
        & (
            pd.to_datetime(df["date"])
            == fight_date
        )
    ]

    if source.empty:
        print("Source fight not found.")
        continue

    fight = source.iloc[0]

    print("\nCURRENT FIGHT ROW")
    print("-" * 80)
    print(
        fight[
            [
                "R_fighter",
                "B_fighter",
                "date",
                "R_Reach_cms",
                "B_Reach_cms",
            ]
        ]
    )

    red_history = history_engine.get_history(
        fighter=red,
        before_date=fight_date,
    )

    blue_history = history_engine.get_history(
        fighter=blue,
        before_date=fight_date,
    )

    red_profile = physical_engine.build(
        history=red_history,
        fighter=red,
        fallback_fight=fight,
    )

    blue_profile = physical_engine.build(
        history=blue_history,
        fighter=blue,
        fallback_fight=fight,
    )

    print("\nPROFILES")
    print("-" * 80)
    print("Red :", red_profile)
    print("Blue:", blue_profile)

    print("\nEXPECTED REACH DIFFERENCE")
    print("-" * 80)

    if pd.isna(red_profile.reach) or pd.isna(
        blue_profile.reach
    ):
        print("NaN — one reach is unavailable")
    else:
        print(
            red_profile.reach
            - blue_profile.reach
        )
```
