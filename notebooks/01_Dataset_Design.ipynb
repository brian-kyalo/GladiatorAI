# Objective

The purpose of this notebook is to design the dataset that will power Gladiator AI.

• the prediction problem

• the unit of observation

• the target variable

• the features

• possible sources of bias

This notebook acts as the blueprint for the entire machine learning pipeline.

# Background

Mixed Martial Arts is one of the most statistically rich sports.

Every fight generates hundreds of measurable variables including:

- striking

- grappling

- defence

- physical attributes

- experience

- recent form

- opponent quality

However, predicting the winner of a fight is difficult because many variables interact simultaneously.

The purpose of Gladiator AI is not simply to predict winners.

Instead it seeks to build an explainable machine learning system capable of analysing historical information and estimating the probability of victory while explaining why the prediction was made.

# Problem Statement

Given two UFC fighters before a scheduled fight,

predict the probability that Fighter A wins using only information available before the fight occurs.
