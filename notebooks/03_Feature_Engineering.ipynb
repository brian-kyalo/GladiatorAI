```python
import pandas as pd

df = pd.read_csv("../datasets/raw/ufc-master.csv")
```

```python
gladiator = pd.DataFrame()
```

```python
gladiator["winner"] = df["Winner"]
```

```python
gladiator["fighter_red"] = df["R_fighter"]
gladiator["fighter_blue"] = df["B_fighter"]
```

```python
gladiator["age_diff"] = df["R_age"] - df["B_age"]

gladiator["height_diff"] = df["R_Height_cms"] - df["B_Height_cms"]

gladiator["reach_diff"] = df["R_Reach_cms"] - df["B_Reach_cms"]
```

```python
gladiator["sig_strike_diff"] = (
    df["R_avg_SIG_STR_landed"] -
    df["B_avg_SIG_STR_landed"]
)

gladiator["sig_accuracy_diff"] = (
    df["R_avg_SIG_STR_pct"] -
    df["B_avg_SIG_STR_pct"]
)
```

```python
gladiator["td_diff"] = (
    df["R_avg_TD_landed"] -
    df["B_avg_TD_landed"]
)

gladiator["td_accuracy_diff"] = (
    df["R_avg_TD_pct"] -
    df["B_avg_TD_pct"]
)

gladiator["submission_diff"] = (
    df["R_avg_SUB_ATT"] -
    df["B_avg_SUB_ATT"]
)
```

```python
gladiator["wins_diff"] = (
    df["R_wins"] -
    df["B_wins"]
)

gladiator["losses_diff"] = (
    df["R_losses"] -
    df["B_losses"]
)

gladiator["win_streak_diff"] = (
    df["R_current_win_streak"] -
    df["B_current_win_streak"]
)

gladiator["lose_streak_diff"] = (
    df["R_current_lose_streak"] -
    df["B_current_lose_streak"]
)
```

```python
gladiator["strike_defense_diff"] = (
    df["R_avg_SIG_STR_pct"] -
    df["B_avg_SIG_STR_pct"]
)

gladiator["td_defense_diff"] = (
    df["R_avg_TD_pct"] -
    df["B_avg_TD_pct"]
)
```

```python
gladiator["title_fight"] = df["title_bout"]

gladiator["rounds"] = df["no_of_rounds"]
```

```python
gladiator.to_csv(
    "../datasets/processed/gladiator_v1.csv",
    index=False
)
```
