```python
def scout_card(name):
    red = df[df["R_fighter"] == name]
    blue = df[df["B_fighter"] == name]

    total_fights = len(red) + len(blue)


    

    print("=" * 40)
    print(name.upper())
    print("=" * 40)
    print(f"Total UFC Fights : {total_fights}")
    print(f"Red Corner       : {len(red)}")
    print(f"Blue Corner      : {len(blue)}")

    red_wins = len(red[red["Winner"] == "Red"])
    blue_wins = len(blue[blue["Winner"] == "Blue"])

    wins = red_wins + blue_wins
    losses = total_fights - wins

    print(f"Wins             : {wins}")
    print(f"Losses           : {losses}")
```

```python
scout_card("Israel Adesanya")
```
