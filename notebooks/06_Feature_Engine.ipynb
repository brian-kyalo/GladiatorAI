```python
print("Fighter:", profile.fighter)

print()

print("Record")
print("Wins:", profile.wins)
print("Losses:", profile.losses)
print("Win Rate:", profile.win_rate)

print()

print("Momentum")
print("Win Streak:", profile.current_win_streak)
print("Lose Streak:", profile.current_lose_streak)
print("Momentum Score:", profile.momentum_score)

print()

print("Finishing")
print("KO Wins:", profile.ko_wins)
print("Submission Wins:", profile.submission_wins)
print("Decision Wins:", profile.decision_wins)

print()

print("Performance")
print("Average Fight Time:", profile.average_fight_time)
print("Average Significant Strikes:", profile.average_sig_strikes)
print("Average Takedowns:", profile.average_takedowns)
print("Average Submission Attempts:", profile.average_submissions)

print()

print("Experience")
print("Experience Score:", profile.experience_score)
```
